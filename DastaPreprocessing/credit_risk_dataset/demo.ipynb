```python
import pandas as pd 
import numpy as np
from sklearn.preprocessing import LabelEncoder ,StandardScaler,MinMaxScaler

print("Successfully imported")
```

```python
df=pd.read_csv("credit_risk_dset.csv")
print("original Dataset")
print(df)
```

```python
print("\nDataset Information")
print(df.info())
```

```python
print("\nMissing Values")
print(df.isnull().sum())
```

```python
df.fillna(df.mean(numeric_only=True), inplace=True)
df.drop_duplicates(inplace=True)
```

```python
print("Mean")
print(np.mean(df[["age","annual_income","account_age_months","avg_deposit_amount","debit_card_spending","credit_score"]]))
```

```python
print("SD")
print(np.std(df[["age","annual_income","account_age_months","avg_deposit_amount","debit_card_spending","credit_score"]]))
```

```python
le = LabelEncoder()
df["gender"] = le.fit_transform(df["gender"])
df["employment_status"] = le.fit_transform(df["employment_status"])
```

# Standardscaler

```python
df_standard = df.copy()

standard = StandardScaler()
df_standard[["age","annual_income","account_age_months","avg_monthly_balance","avg_deposit_amount","debit_card_spending","credit_score"]] = standard.fit_transform(
    df_standard[["age","annual_income","account_age_months","avg_monthly_balance","avg_deposit_amount","debit_card_spending","credit_score"]]
)

print("\nDataset after StandardScaler")
print(df_standard)
```

```python
df_minmax = df.copy()

minmax = MinMaxScaler()

df_minmax[["age","annual_income","account_age_months","avg_monthly_balance","avg_deposit_amount","debit_card_spending","credit_score"]] = minmax.fit_transform(
    df_minmax[["age","annual_income","account_age_months","avg_monthly_balance","avg_deposit_amount","debit_card_spending","credit_score"]]
)

print("\nDataset after MinMaxScaler")
print(df_minmax)
```

```python
df_standard.to_excel("standard_scaled.xlsx", index=False)
df_minmax.to_excel("minmax_scaled.xlsx", index=False)

print("\nDatasets Saved Successfully")
```
